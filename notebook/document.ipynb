```python
import os
```

```python
os.makedirs("../data/text_files", exist_ok= True)
```

```python
sample_texts ={
    "../data/text_files/python_intro.txt":"""
    Python is a versatile, high-level programming language created by Guido van Rossum and first released in 1991. Designed with a philosophy that emphasizes code readability, Python uses significant indentation to keep code clean, organized, and easy to understand. This makes it an ideal language for beginners learning to code, as well as seasoned professionals building complex systems.

Why Choose Python?

Readability: Python's syntax closely resembles human language, allowing developers to focus on solving problems rather than struggling with complex syntax rules.

Massive Ecosystem: It comes with a "batteries included" standard library and a massive ecosystem of third-party packages (available through tools like pip and uv), meaning you rarely have to build tools from scratch.

Cross-Platform: Python runs seamlessly across Windows, macOS, and Linux.

What is Python Used For?
Because it is a general-purpose language, Python can be used for almost anything. Some of its most popular applications include:

Data Science and AI: It is the industry standard for machine learning, artificial intelligence, and data analysis, powered by libraries like Pandas, NumPy, and TensorFlow.

Web Development: Frameworks like Django and FastAPI allow developers to build robust, scalable web applications quickly.

Automation and Scripting: Python excels at automating repetitive tasks, managing system operations, and writing quick scripts to process data.

Whether you are automating a simple daily task or building the next breakthrough machine learning model, Python provides the tools, community, and simplicity to help you get it done."""
}

for filepath,content in sample_texts.items():
    with open(filepath,'w', encoding="utf-8") as f:
        f:write(content)
```

```python
from langchain_community.document_loaders import TextLoader
```

```python
loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document=loader.load()
print(document)
```

```python
## Directory Loader

from langchain_community.document_loaders import DirectoryLoader

#load all the text files

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls = TextLoader,
    loader_kwargs={'encoding':'utf-8'},
    show_progress=False
)

documents=dir_loader.load()
documents
```
